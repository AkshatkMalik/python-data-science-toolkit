**Exercise 1: Prime Numbers**

```python
# Get number
num = int(input("Enter a number to check if it's prime: "))

# Prime numbershave to be greater than 1
if num <= 1:
  print(f"{num} is not a prime number.")
else:
  is_prime = True

  # Checking for factors from 2 up to num - 1
  for i in range(2, num):
    if num % i == 0:  # If it is divisible, it's not prime
      is_prime = False
      break

  # Printing the final result
  if is_prime:
    print(f"{num} is a prime number.")
  else:
    print(f"{num} is not a prime number.")
```

**Exercise 2: Product of Random Numbers**

```python
import random

# Generating two random numbers between 1 and 10
num1 = random.randint(1, 10)
num2 = random.randint(1, 10)

# Calculating the actual correct product
correct_product = num1 * num2

# user_input
user_answer = int(input(f"What is the product of {num1} and {num2}? "))

# Check if the user's answer is correct
if user_answer == correct_product:
  print("Correct! Great job!")
else:
  print(f"Incorrect. The correct answer was {correct_product}.")
```

**Exercise 3: Squares of Even/Odd Numbers**

```python
print("Squares of even numbers from 100 to 200:")

# Loop through numbers from 100 to 200 (inclusive)
for num in range(100, 201):
  # Check if the number is even
  if num % 2 == 0:
    square = num**2
    print(f"Square of {num} is {square}")
```

**Exercise 4: Word Counter**

```python
input_text = (
    "This is a sample text. This text will be used to demonstrate the word"
    " counter."
)

# Step 1: Spliting the text into a list of words using spaces
words = input_text.split()

# Step 2: Creating an empty dictionary to store word counts
word_count = {}

# Step 3: Loop through each word and count occurrences
for word in words:
  if word in word_count:
    word_count[word] += 1
  else:
    word_count[word] = 1

# Step 4: Printing the results in the requested format
for word, count in word_count.items():
  print(f"'{word}': {count}")
```

**Exercise 5: Check for Palindrome**

```python
def is_palindrome(text):
  # Step 1: Convertong to lowercase and keep only letters/numbers
  cleaned_text = ""
  for char in text.lower():
    if char.isalnum():
      cleaned_text += char

  # Step 2: Comparing the cleaned text with its reverse
  return cleaned_text == cleaned_text[::-1]


# Example testing
input_str = "racecar"
result = is_palindrome(input_str)

print(f"Input: \"{input_str}\"")
print(f"Expected Output: {result}")
```
